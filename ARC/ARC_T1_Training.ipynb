# ARC T1 Training Notebook

Run the shared lesion segmentation pipeline against the ARC T1-weighted dataset.

```python
from pathlib import Path
import importlib.util

MODULE_PATH = Path("/home/rbielski/stroke_cleaned/stroke_segmentation_v1.2/stroke_seg_v1.2_train.py")
spec = importlib.util.spec_from_file_location("arc_seg_train", MODULE_PATH)
seg = importlib.util.module_from_spec(spec)
spec.loader.exec_module(seg)

history = seg.train_dynamic_model(
    DATA_DIR=Path("/home/rbielski/ARC/ds004884/derivatives/aggregates/t1w_with_masks/mni_1mm_ants"),
    MODEL_DIR=Path("/home/rbielski/stroke_cleaned/ARC/models"),
    CALLBACKS_DIR=Path("/home/rbielski/stroke_cleaned/ARC/callbacks"),
    TOTAL_EPOCHS=60,
    VALIDATION_SPLIT=0.1,
    BATCH_SIZE=2,
    BASE_FILTERS=6,
    RESAMPLE_TO_TARGET=True
)
print("Training complete. History keys:", list(history.history.keys()))
```
