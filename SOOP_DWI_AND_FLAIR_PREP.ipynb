```python
# === Modality shape catalog (DWI vs FLAIR vs acute mask) ======================
from pathlib import Path
import nibabel as nib

DWI_IMAGES_DIR = Path("/home/rbielski/SOOP/ds004889/acute_only/DWI_images/train")
FLAIR_IMAGES_DIR = Path("/home/rbielski/SOOP/ds004889/acute_only/FLAIR_images/train")
MASKS_DIR = Path("/home/rbielski/SOOP/ds004889/acute_only/Acute_mask/train")

def dwi_key(name: str) -> str:
    stem = name[:-7] if name.endswith(".nii.gz") else name
    return stem.split("_rec-TRACE")[0]

def flair_key(name: str) -> str:
    stem = name[:-7] if name.endswith(".nii.gz") else name
    return stem.replace("_FLAIR", "")

def mask_key(name: str) -> str:
    stem = name[:-7] if name.endswith(".nii.gz") else name
    return stem.split("_space-TRACE")[0]

dwi_map = {dwi_key(p.name): p for p in sorted(DWI_IMAGES_DIR.glob("*.nii.gz")) if "mask" not in p.name}
flair_map = {flair_key(p.name): p for p in sorted(FLAIR_IMAGES_DIR.glob("*.nii.gz")) if "mask" not in p.name}
mask_map = {mask_key(p.name): p for p in sorted(MASKS_DIR.glob("*.nii.gz")) if "mask" in p.name}

common_keys = sorted(set(dwi_map) & set(flair_map) & set(mask_map))
print(f"Total subjects with DWI+FLAIR+acute mask: {len(common_keys)}")



catalog = []
for key in common_keys:
    dwi_shape = nib.load(str(dwi_map[key])).shape[:3]
    flair_shape = nib.load(str(flair_map[key])).shape[:3]
    mask_shape = nib.load(str(mask_map[key])).shape[:3]
    flair_voxels = int(flair_shape[0] * flair_shape[1] * flair_shape[2])
    catalog.append((key, dwi_shape, flair_shape, mask_shape, flair_voxels))

catalog.sort(key=lambda row: row[-1], reverse=True)
for idx, (key, dwi_shape, flair_shape, mask_shape, voxels) in enumerate(catalog, start=1):
    print(f"{idx:03d}. {key} -> DWI {dwi_shape}, FLAIR {flair_shape} ({voxels:,} voxels), mask {mask_shape}")
```
