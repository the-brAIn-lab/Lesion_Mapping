# Stroke Outcome Optimization Project (SOOP)

## Overview

This dataset includes anonymized images, behavioral measures and demographic details from a cohort of individuals from South Carolina with acute stroke. The fully BIDS-compatible dataset is fully anonymized, allowing public sharing which is vital for education and development of BIDS pipelines that are capable of processing clinical datasets. Along with its companion chronic [Aphasia Recovery Cohort (ARC)](https://openneuro.org/datasets/ds004512/), these features complement existing repositories of [acute](https://www.ncbi.nlm.nih.gov/pmc/articles/PMC10444746/ and [chronic](https://pubmed.ncbi.nlm.nih.gov/32310331/) stroke.

## Data Contents
* `participants.tsv` - List of subject IDs, demographic variables, and impairment measures. Note that many variables are [generalized](https://www.hhs.gov/hipaa/for-professionals/privacy/special-topics/de-identification/index.html#approachmitigate) to meet [International Safe Harbor Privacy Principles](https://en.wikipedia.org/wiki/International_Safe_Harbor_Privacy_Principles).


* `Subject Directories` - MRI data directories for each subject. Note that sequence protocols and diffusion derivative formula vary between individuals. See the individual BIDS JSON sidecars included with each image for specific details.
  - `anat` - T1w and T2w-FLAIR anatomical images
  - `dwi` - Diffusion weighted Trace and  apparent diffusion coefficient (ADC) images

### Exploratory Data Analysis: Stroke Status & Modalities

This section classifies subjects into stroke status categories using `participants.tsv` and audits imaging modality availability.

Classification logic (per subject):
- `acute_only`: acuteischaemicstroke = 1, priorstroke = 0
- `chronic_only`: acuteischaemicstroke = 0, priorstroke = 1
- `acute+chronic`: acuteischaemicstroke = 1, priorstroke = 1
- `no_stroke_reported`: acuteischaemicstroke = 0, priorstroke = 0
- `partial_unknown`: one of the two fields missing / n/a
- `unknown`: both fields missing / n/a

Modalities checked in raw BIDS folders:
- T1w (`anat/*_T1w.nii.gz`)
- FLAIR (`anat/*_FLAIR.nii.gz`)
- DWI TRACE (`dwi/*_TRACE_dwi.nii.gz`)
- DWI ADC (`dwi/*_ADC_dwi.nii.gz`)

Lesion mask derivatives (if present):
- Acute mask: `desc-lesionAcute_mask`
- Chronic mask: `desc-lesionChronic_mask`
- Combined mask: `desc-lesion_mask`

Outputs below:
1. Stroke status counts & percentages.
2. Cross-tab of stroke status vs lesion mask availability.
3. Modality coverage summary (how many subjects have each modality).
4. Any subjects present in filesystem but absent from participants file (and vice versa).


```python
# Modality & Lesion Mask Coverage Analysis (robust / self-contained)
import pandas as pd, re, unicodedata
from pathlib import Path

# Reconstruct base and (re)load participants with classification if prior cell not run
base = Path('ds004889')
participants_path = base / 'participants.tsv'  # use relative BIDS root
na_vals = ['n/a', 'NA', 'NaN', '', 'NULL', 'null']

def _clean_col(c: str) -> str:
    c = unicodedata.normalize('NFKC', c).replace('\ufeff','')
    return re.sub(r'\s+', '_', c.strip().lower())

if 'df' not in globals() or 'stroke_status' not in globals().get('df', pd.DataFrame()).columns:
    df = pd.read_csv(participants_path, sep='\t', na_values=na_vals, dtype=str)
    orig_cols = list(df.columns)
    df.columns = [_clean_col(c) for c in df.columns]
    print(f"Original columns: {orig_cols}\nNormalized columns: {list(df.columns)}")

    # Detect participant id column robustly
    pid_col = None
    for c in df.columns:
        if c.startswith('participant'):
            pid_col = c; break
    if pid_col and pid_col != 'participant_id':
        df.rename(columns={pid_col: 'participant_id'}, inplace=True)
        print(f"✓ Renamed '{pid_col}' -> 'participant_id'")

    if 'participant_id' not in df.columns:
        print("⚠️ 'participant_id' column missing after normalization; creating from index.")
        df['participant_id'] = [f'sub-idx{i}' for i in range(len(df))]

    for c in ['acuteischaemicstroke', 'priorstroke']:
        if c in df.columns:
            df[c] = pd.to_numeric(df[c], errors='coerce')

    def classify(row):
        a = row.get('acuteischaemicstroke'); p = row.get('priorstroke')
        if pd.isna(a) and pd.isna(p): return 'unknown'
        if pd.isna(a) or pd.isna(p): return 'partial_unknown'
        if a == 1 and p == 1: return 'acute+chronic'
        if a == 1 and p == 0: return 'acute_only'
        if a == 0 and p == 1: return 'chronic_only'
        if a == 0 and p == 0: return 'no_stroke_reported'
        return 'unclassified'

    if 'stroke_status' not in df.columns:
        df['stroke_status'] = df.apply(classify, axis=1)

# ---------------------- new defensive guard (patch) ----------------------
# Ensure both required columns exist; if not, force a clean reload
required = {'participant_id','stroke_status'}
missing_now = [c for c in required if c not in df.columns]

if missing_now:
    print(f"♻️ Re-loading participants.tsv because missing columns: {missing_now}")
    df_reload = pd.read_csv(participants_path, sep='\t', na_values=na_vals, dtype=str)
    df_reload.columns = [_clean_col(c) for c in df_reload.columns]

    # Rebuild / rename participant id
    if 'participant_id' not in df_reload.columns:
        candidate = next((c for c in df_reload.columns if c.startswith('participant')), None)
        if candidate and candidate != 'participant_id':
            df_reload.rename(columns={candidate: 'participant_id'}, inplace=True)
    if 'participant_id' not in df_reload.columns:
        df_reload['participant_id'] = [f'sub-idx{i}' for i in range(len(df_reload))]

    # Stroke columns
    for c in ['acuteischaemicstroke','priorstroke']:
        if c in df_reload.columns:
            df_reload[c] = pd.to_numeric(df_reload[c], errors='coerce')

    if 'stroke_status' not in df_reload.columns:
        df_reload['stroke_status'] = df_reload.apply(classify, axis=1)

    # Use reloaded frame only if it now satisfies
    if required.issubset(df_reload.columns):
        df = df_reload
        print("✅ Successfully reconstructed missing columns.")
    else:
        # fallback: synthesize minimal columns
        print("⚠️ Could not fully reconstruct; synthesizing placeholders.")
        if 'participant_id' not in df.columns:
            df['participant_id'] = [f'sub-idx{i}' for i in range(len(df))]
        if 'stroke_status' not in df.columns:
            df['stroke_status'] = 'unknown'

# Final sanity
req_cols = {'participant_id','stroke_status'}
missing_req = req_cols - set(df.columns)
if missing_req:
    print(f"❌ Required columns still missing: {missing_req} (will fallback).")

# ---------------------- existing filesystem scan code (unchanged) ----------------------
subjects_fs = sorted([p.name for p in base.glob('sub-*')
                      if p.is_dir() and re.match(r'sub-\d+', p.name)])

mask_stats = {
    'acute_found': 0, 'chronic_found': 0, 'combined_found': 0,
    'subjects_with_acute': set(), 'subjects_with_chronic': set(),
    'subjects_with_combined': set(), 'subjects_with_any_mask': set()
}

records = []
for sid in subjects_fs:
    subj_dir = base / sid
    anat = subj_dir / 'anat'; dwi = subj_dir / 'dwi'
    has_t1w = any(anat.glob(f'{sid}_T1w.nii.gz')) if anat.exists() else False
    has_flair = any(anat.glob(f'{sid}_FLAIR.nii.gz')) if anat.exists() else False
    has_trace = any(dwi.glob(f'{sid}_rec-TRACE_dwi.nii.gz')) if dwi.exists() else False
    has_adc = any(dwi.glob(f'{sid}_rec-ADC_dwi.nii.gz')) if dwi.exists() else False
    mask_dir = base / 'derivatives' / 'lesion_masks' / sid / 'dwi'
    has_acute_mask = any(mask_dir.glob(f'{sid}_space-TRACE_desc-lesionAcute_mask.nii.gz')) if mask_dir.exists() else False
    has_chronic_mask = any(mask_dir.glob(f'{sid}_space-TRACE_desc-lesionChronic_mask.nii.gz')) if mask_dir.exists() else False
    has_combined_mask = any(mask_dir.glob(f'{sid}_space-TRACE_desc-lesion_mask.nii.gz')) if mask_dir.exists() else False
    if has_acute_mask:
        mask_stats['acute_found'] += 1; mask_stats['subjects_with_acute'].add(sid); mask_stats['subjects_with_any_mask'].add(sid)
    if has_chronic_mask:
        mask_stats['chronic_found'] += 1; mask_stats['subjects_with_chronic'].add(sid); mask_stats['subjects_with_any_mask'].add(sid)
    if has_combined_mask:
        mask_stats['combined_found'] += 1; mask_stats['subjects_with_combined'].add(sid); mask_stats['subjects_with_any_mask'].add(sid)
    records.append({
        'participant_id': sid, 'has_T1w': has_t1w, 'has_FLAIR': has_flair,
        'has_TRACE': has_trace, 'has_ADC': has_adc,
        'has_acute_mask': has_acute_mask, 'has_chronic_mask': has_chronic_mask,
        'has_combined_mask': has_combined_mask
    })

mod_df = pd.DataFrame(records)

# ---------------------- patched safe merge ----------------------
if req_cols.issubset(df.columns):
    try:
        merged = mod_df.merge(df[['participant_id','stroke_status']], on='participant_id', how='left')
    except KeyError as e:
        print(f"⚠️ Merge KeyError: {e}. Forcing fallback classification = 'unknown'.")
        merged = mod_df.copy()
        merged['stroke_status'] = 'unknown'
else:
    merged = mod_df.copy()
    merged['stroke_status'] = 'unknown'

acute_chronic_count = (merged['stroke_status'] == 'acute+chronic').sum()
chronic_only_count = (merged['stroke_status'] == 'chronic_only').sum()

print('\n' + '='*60)
print('CLINICAL STROKE STATUS BREAKDOWN:')
print('='*60)
print(f"🔥 Patients with ACUTE + CHRONIC stroke: {acute_chronic_count}")
print(f"🟡 Patients with CHRONIC ONLY stroke:    {chronic_only_count}")

print(f"\n📊 MASK COUNTS BY TYPE:")
print(f"   🔴 Acute masks:     {mask_stats['acute_found']:4} files")
print(f"   🟡 Chronic masks:   {mask_stats['chronic_found']:4} files")
print(f"   🔵 Combined masks:  {mask_stats['combined_found']:4} files")

print('\n✅ Merge completed with participant_id' if not missing_req else '\n✅ Merge completed (fallback used)')
```

#### Interpretation Notes & Next Steps
- Percentages for modalities are relative to subjects with folders, not rows in `participants.tsv`.
- Missing clinical rows (if any) should be investigated—may indicate subjects scanned but not fully annotated.
- High rate of `partial_unknown` suggests incomplete abstraction; consider excluding or imputing carefully.
- For modeling acute lesion segmentation, you can subset: `stroke_status in ['acute_only','acute+chronic']` AND `has_acute_mask`.
- For chronic-only analyses: `stroke_status in ['chronic_only','acute+chronic']` AND `has_chronic_mask`.
- Validate consistency: a subject labeled `acute_only` should not have a chronic mask; add QC if discrepancies appear.

### Exploratory Data Analysis: Stroke Status & Modalities

This section classifies subjects into stroke status categories using `participants.tsv` and audits imaging modality availability.

Classification logic (per subject):
- `acute_only`: acuteischaemicstroke = 1, priorstroke = 0
- `chronic_only`: acuteischaemicstroke = 0, priorstroke = 1
- `acute+chronic`: acuteischaemicstroke = 1, priorstroke = 1
- `no_stroke_reported`: acuteischaemicstroke = 0, priorstroke = 0
- `partial_unknown`: one of the two fields missing / n/a
- `unknown`: both fields missing / n/a

Modalities checked in raw BIDS folders:
- T1w (`anat/*_T1w.nii.gz`)
- FLAIR (`anat/*_FLAIR.nii.gz`)
- DWI TRACE (`dwi/*_TRACE_dwi.nii.gz`)
- DWI ADC (`dwi/*_ADC_dwi.nii.gz`)

Lesion mask derivatives (if present):
- Acute mask: `desc-lesionAcute_mask`
- Chronic mask: `desc-lesionChronic_mask`
- Combined mask: `desc-lesion_mask`

Outputs below:
1. Stroke status counts & percentages.
2. Cross-tab of stroke status vs lesion mask availability.
3. Modality coverage summary (how many subjects have each modality).
4. Any subjects present in filesystem but absent from participants file (and vice versa).


```python
# Standardize All Mask Dimensions to Match DWI Images (ACUTE + CHRONIC folder only)
import nibabel as nib
import numpy as np
from pathlib import Path
import shutil
from tqdm import tqdm

# Paths to ACUTE+CHRONIC organized dataset
base_path = Path('/home/rbielski/SOOP/ds004889/acute_chronic')
dwi_dir = base_path / 'DWI_images'
acute_dir = base_path / 'Acute_mask'
chronic_dir = base_path / 'Chronic_mask'
combined_dir = base_path / 'Combined_mask'

def normalize_to_3d(data):
    """Ensure data is 3D by squeezing extra dimensions"""
    data = np.squeeze(data)
    if data.ndim == 4:
        # Take first volume if 4D
        data = data[:, :, :, 0]
    elif data.ndim == 2:
        # Add slice dimension if 2D
        data = data[:, :, np.newaxis]
    return data

def resize_to_match(mask_data, target_shape):
    """Resize mask to match target shape using cropping/padding"""
    if mask_data is None:
        return None
    
    current_shape = mask_data.shape
    
    # Handle each dimension
    resized = mask_data.copy()
    
    for dim in range(3):
        current_size = current_shape[dim]
        target_size = target_shape[dim]
        
        if current_size == target_size:
            continue
        elif current_size > target_size:
            # Crop from center
            start = (current_size - target_size) // 2
            if dim == 0:
                resized = resized[start:start + target_size, :, :]
            elif dim == 1:
                resized = resized[:, start:start + target_size, :]
            else:  # dim == 2
                resized = resized[:, :, start:start + target_size]
        else:
            # Pad symmetrically
            pad_total = target_size - current_size
            pad_before = pad_total // 2
            pad_after = pad_total - pad_before
            
            pad_width = [(0, 0), (0, 0), (0, 0)]
            pad_width[dim] = (pad_before, pad_after)
            resized = np.pad(resized, pad_width, mode='constant', constant_values=0)
    
    return resized

def process_all_masks():
    """Process all masks in ACUTE+CHRONIC folder to match their corresponding DWI dimensions"""
    
    # Get all DWI files from acute+chronic folder
    dwi_files = list(dwi_dir.glob('*_rec-TRACE_dwi.nii.gz'))
    
    print(f"Processing {len(dwi_files)} DWI images and their masks in ACUTE+CHRONIC folder...")
    
    stats = {
        'processed': 0,
        'acute_fixed': 0,
        'chronic_fixed': 0,
        'combined_fixed': 0,
        'errors': []
    }
    
    for dwi_file in tqdm(dwi_files, desc="Processing acute+chronic subjects"):
        subject_id = dwi_file.name.split('_rec-TRACE')[0]
        
        try:
            # Load DWI to get target dimensions
            dwi_nii = nib.load(str(dwi_file))
            dwi_data = dwi_nii.get_fdata()
            dwi_normalized = normalize_to_3d(dwi_data)
            target_shape = dwi_normalized.shape
            
            # Process each mask type
            mask_types = ['acute', 'chronic', 'combined']
            mask_dirs = [acute_dir, chronic_dir, combined_dir]
            mask_patterns = [
                f'{subject_id}_space-TRACE_desc-lesionAcute_mask.nii.gz',
                f'{subject_id}_space-TRACE_desc-lesionChronic_mask.nii.gz',
                f'{subject_id}_space-TRACE_desc-lesion_mask.nii.gz'
            ]
            
            for mask_type, mask_dir, pattern in zip(mask_types, mask_dirs, mask_patterns):
                mask_file = mask_dir / pattern
                
                if mask_file.exists():
                    # Load original mask
                    mask_nii = nib.load(str(mask_file))
                    mask_data = mask_nii.get_fdata()
                    mask_normalized = normalize_to_3d(mask_data)
                    
                    # Check if dimensions match
                    if mask_normalized.shape != target_shape:
                        # Resize mask to match DWI
                        mask_resized = resize_to_match(mask_normalized, target_shape)
                        
                        # Create new NIfTI with DWI header (same space/orientation)
                        mask_resized_nii = nib.Nifti1Image(
                            mask_resized.astype(mask_data.dtype),
                            dwi_nii.affine,
                            dwi_nii.header
                        )
                        
                        # Save the resized mask
                        nib.save(mask_resized_nii, str(mask_file))
                        stats[f'{mask_type}_fixed'] += 1
                        
                        print(f"✓ {subject_id} {mask_type}: {mask_normalized.shape} → {target_shape}")
            
            stats['processed'] += 1
            
        except Exception as e:
            error_msg = f"Error processing {subject_id}: {str(e)}"
            stats['errors'].append(error_msg)
            print(f"❌ {error_msg}")
    
    return stats

# Run the preprocessing for acute+chronic folder only
print("🔧 Starting mask dimension standardization for ACUTE+CHRONIC folder...")
print("This will modify masks in ACUTE+CHRONIC folder to match DWI dimensions exactly.\n")

results = process_all_masks()

print(f"\n" + "="*60)
print("PREPROCESSING RESULTS (ACUTE+CHRONIC):")
print("="*60)
print(f"✅ Subjects processed: {results['processed']}")
print(f"🔧 Acute masks fixed: {results['acute_fixed']}")
print(f"🔧 Chronic masks fixed: {results['chronic_fixed']}")
print(f"🔧 Combined masks fixed: {results['combined_fixed']}")
print(f"❌ Errors encountered: {len(results['errors'])}")

if results['errors']:
    print("\nErrors:")
    for error in results['errors'][:5]:  # Show first 5 errors
        print(f"  - {error}")
    if len(results['errors']) > 5:
        print(f"  ... and {len(results['errors']) - 5} more")

print(f"\n🎯 All acute+chronic masks now have matching dimensions with their DWI images!")
```

### Acute vs General Mask: Interpretation

- In your example (`sub-4`), the acute mask and general mask are **identical**: same shape, same number of nonzero voxels, 100% overlap.
- This means for this subject, the general mask (`desc-lesion_mask.nii.gz`) is just a copy of the acute mask (`desc-lesionAcute_mask.nii.gz`).
- In other subjects, the general mask may include both acute and chronic lesions, but for `acute_only` subjects, they are often the same.

**Summary:**
- For acute_only patients, the general mask is usually just the acute mask.
- For patients with both acute and chronic stroke, the general mask may be the union of acute and chronic masks.

# Stroke Outcome Optimization Project (SOOP)

## Overview

This dataset includes anonymized images, behavioral measures and demographic details from a cohort of individuals from South Carolina with acute stroke. The fully BIDS-compatible dataset is fully anonymized, allowing public sharing which is vital for education and development of BIDS pipelines that are capable of processing clinical datasets. Along with its companion chronic [Aphasia Recovery Cohort (ARC)](https://openneuro.org/datasets/ds004512/), these features complement existing repositories of [acute](https://www.ncbi.nlm.nih.gov/pmc/articles/PMC10444746/ and [chronic](https://pubmed.ncbi.nlm.nih.gov/32310331/) stroke.

## Data Contents
* `participants.tsv` - List of subject IDs, demographic variables, and impairment measures. Note that many variables are [generalized](https://www.hhs.gov/hipaa/for-professionals/privacy/special-topics/de-identification/index.html#approachmitigate) to meet [International Safe Harbor Privacy Principles](https://en.wikipedia.org/wiki/International_Safe_Harbor_Privacy_Principles).


* `Subject Directories` - MRI data directories for each subject. Note that sequence protocols and diffusion derivative formula vary between individuals. See the individual BIDS JSON sidecars included with each image for specific details.
  - `anat` - T1w and T2w-FLAIR anatomical images
  - `dwi` - Diffusion weighted Trace and  apparent diffusion coefficient (ADC) images

### Exploratory Data Analysis: Stroke Status & Modalities

This section classifies subjects into stroke status categories using `participants.tsv` and audits imaging modality availability.

Classification logic (per subject):
- `acute_only`: acuteischaemicstroke = 1, priorstroke = 0
- `chronic_only`: acuteischaemicstroke = 0, priorstroke = 1
- `acute+chronic`: acuteischaemicstroke = 1, priorstroke = 1
- `no_stroke_reported`: acuteischaemicstroke = 0, priorstroke = 0
- `partial_unknown`: one of the two fields missing / n/a
- `unknown`: both fields missing / n/a

Modalities checked in raw BIDS folders:
- T1w (`anat/*_T1w.nii.gz`)
- FLAIR (`anat/*_FLAIR.nii.gz`)
- DWI TRACE (`dwi/*_TRACE_dwi.nii.gz`)
- DWI ADC (`dwi/*_ADC_dwi.nii.gz`)

Lesion mask derivatives (if present):
- Acute mask: `desc-lesionAcute_mask`
- Chronic mask: `desc-lesionChronic_mask`
- Combined mask: `desc-lesion_mask`

Outputs below:
1. Stroke status counts & percentages.
2. Cross-tab of stroke status vs lesion mask availability.
3. Modality coverage summary (how many subjects have each modality).
4. Any subjects present in filesystem but absent from participants file (and vice versa).


```python
# === FLAIR + mask shape catalog ==============================================
from pathlib import Path
import nibabel as nib

IMAGES_DIR = Path("/home/rbielski/SOOP/ds004889/acute_only/DWI_images/train")
MASKS_DIR = Path("/home/rbielski/SOOP/ds004889/acute_only/Acute_mask/train")

def _img_key(name: str) -> str:
    stem = name[:-7] if name.endswith(".nii.gz") else name
    return stem.split("_rec-TRACE")[0]

def _msk_key(name: str) -> str:
    stem = name[:-7] if name.endswith(".nii.gz") else name
    return stem.split("_space-TRACE")[0]

images = { _img_key(p.name): p for p in sorted(IMAGES_DIR.glob("*.nii.gz")) if "mask" not in p.name }
masks  = { _msk_key(p.name): p for p in sorted(MASKS_DIR.glob("*.nii.gz")) if "mask" in p.name }

pairs = sorted(set(images) & set(masks))
catalog = []
for key in pairs:
    img_path = images[key]
    msk_path = masks[key]
    img_shape = nib.load(str(img_path)).shape[:3]
    msk_shape = nib.load(str(msk_path)).shape[:3]
    img_voxels = int(img_shape[0] * img_shape[1] * img_shape[2])
    catalog.append((key, img_path.name, img_shape, msk_shape, img_voxels))

catalog.sort(key=lambda entry: entry[-1], reverse=True)
print(f"Total paired cases: {len(catalog)}")
for idx, (_, name, img_shape, msk_shape, voxels) in enumerate(catalog, start=1):
    print(f"{idx:03d}. {name} -> image shape {img_shape} ({voxels:,} voxels), mask shape {msk_shape}")
```
